```python
from opcua import Server, ua
import time
import random

# ====================== 1. 创建 OPC UA 服务器 ======================
server = Server()

# 服务器地址（0.0.0.0 允许所有设备连接）
url = "opc.tcp://0.0.0.0:4840"
server.set_endpoint(url)

# 服务器名称
server.set_server_name("Python Simulated PLC OPC UA Server")
server.set_application_uri("urn:python:simulated:plc")

# ====================== 2. 建立 PLC 地址空间结构 ======================
# 根节点
objects = server.get_objects_node()

# 创建 PLC 主节点
plc = objects.add_object("ns=2;s=PLC", "Simulated_PLC")

# 创建 4 个 PLC 变量组
di = plc.add_object("ns=2;s=DI", "Digital_Inputs")   # 数字输入
do = plc.add_object("ns=2;s=DO", "Digital_Outputs")  # 数字输出
ai = plc.add_object("ns=2;s=AI", "Analog_Inputs")    # 模拟输入
ao = plc.add_object("ns=2;s=AO", "Analog_Outputs")   # 模拟输出

# ====================== 3. 添加 PLC 点位 ======================
# 数字量 DI/DO（True/False）
di1 = di.add_variable("ns=2;s=DI1", "DI1", False)
di2 = di.add_variable("ns=2;s=DI2", "DI2", True)
do1 = do.add_variable("ns=2;s=DO1", "DO1", False)
do2 = do.add_variable("ns=2;s=DO2", "DO2", True)

# 模拟量 AI/AO（浮点数）
ai1 = ai.add_variable("ns=2;s=AI1", "Temperature", 25.0)  # 温度
ai2 = ai.add_variable("ns=2;s=AI2", "Pressure", 1.0)       # 压力
ai3 = ai.add_variable("ns=2;s=AI3", "Flow", 10.0)         # 流量
ao1 = ao.add_variable("ns=2;s=AO1", "AO1", 0.0)

# 设置变量可写（客户端可以修改 DO/AO）
do1.set_writable()
do2.set_writable()
ao1.set_writable()

# ====================== 4. 启动服务器 ======================
print(f"OPC UA 服务器启动：{url}")
print("命名空间索引：2")
print("按 Ctrl+C 停止服务器\n")
server.start()

try:
    while True:
        # 模拟温度自动波动 20~30℃
        #ai1.set_value(round(20 + random.random() * 10, 2))
        ai1.set_value(round(56.78, 2))
        # 模拟压力波动 0.8~1.2 MPa
        ai2.set_value(round(0.8 + random.random() * 0.4, 2))
        
        # 模拟流量波动 8~12 m³/h
        ai3.set_value(round(8 + random.random() * 4, 2))
        
        time.sleep(1)

finally:
    server.stop()
```
